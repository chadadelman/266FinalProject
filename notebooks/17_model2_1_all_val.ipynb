## Run model 2_1 on all validation to see if it is better than the newer model

```python
import torch
from torch.utils.data import DataLoader

def run_model(model
              ,tokenizer
              ,device
              ,test_text_list
              ,max_length = 32
              ,num_beams = 2):
    
    generated_text_list=[]

    for test_text in test_text_list:
        #summarize each item in test_text_list
        inputs = tokenizer(
            "summarize: " + test_text,
            return_tensors="pt",
            truncation=True,
            padding=True,
        ).to(device)

        with torch.no_grad():
            generated_ids = model.generate(
                **inputs,
                max_length=max_length,
                num_beams=num_beams,
            )

        generated_text = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
        generated_text_list.append(generated_text)
    
    return generated_text_list
```
