## Process ChatGPT Summary Text File

```python
with open("../data/summaries/chatgpt.txt", "r") as f:
    text = f.read()
```

```python
#breaking the text file into a dictionary
summaries_batch3={}

summary_list = text.split("\n\n")
for e in summary_list:
    split_summary = e.split("\n")
    k = split_summary[0]
    v = split_summary[1]
    summaries_batch3[k]=v
```

```python
#Fro checking
print(list(summaries_batch3.values())[10])
```
