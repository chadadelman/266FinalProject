```python
#We want human generated summaries to be the validation and test set
#We want the sonnet vs scene manual summary to be split across test vs validation dataset

#Validation Dataset (All manually generated summaries)
#Othello
#Hamlet
#Romeo and Juliette
#Sonnets - odds

#Test Dataset (All manually generated summaries)
#MacBeth
#Julius Caesar
#Twelfth Night
#Sonnets - even

#Training Set (All ChatGPT generated summaries)
#All else

#Stores the names of plays in both dataset, these are used when performing the split
val_dataset_play_names = ["Othello","Hamlet","Romeo_and_Juliet"]
test_dataset_play_names = ["MacBeth","Julius_Caesar","Twelfth_Night"]
```

```python
#Split the database paired_summaries into the following three datasets:
#train_paired_summaries, val_paired_summaries, train_paired_summaries
#These will all have the same format as paired_summaries
#Dictinory with the key being the text name and the value being a 3 element tuple of (text, summary, human-made flag) where human-made flag is true if the summary was made by a human

test_paired_summaries = dict()
val_paired_summaries = dict()
train_paired_summaries = dict()

for k,v in paired_summaries.items():
    
    #If its a sonnet, assign to train or val based on odd or even number
    even_sonnet=False
    odd_sonnet=False
    if "Sonnet" in k:
        sonnet_num = int(k.split(" ")[1])
        if sonnet_num%2==0:
            even_sonnet=True
        else:
            odd_sonnet=True
    
    #add the element to the appropriate train or val dataset
    if any(item in k for item in val_dataset_play_names) or odd_sonnet:
        #scenes in val dataset
        val_paired_summaries[k]=v
    
    elif any(item in k for item in test_dataset_play_names) or even_sonnet:
        #scenes in test dataset
        test_paired_summaries[k]=v

    #not validation or test so is thus train
    else:
        train_paired_summaries[k]=v
    
```

```python
#Looking throught data
print("Train Dataset Length:",len(train_paired_summaries))
print("Val Dataset Length:",len(val_paired_summaries))
print("Test Dataset Length:",len(test_paired_summaries))
```
