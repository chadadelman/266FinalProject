```python
#Pairing of scenes to summaries
#Will be dictinory with the key being the text name and the value being a 3 element tuple of (text, summary, human-made flag) where human-made flag is true if the summary was made by a human
paired_summaries = {}

for k,v in all_text_dict.items():
    if k in summaries_batch3: #made by ChatGPT
        human_made = False
    else:
        human_made = True
    
    the_summary = all_summaries_dict[k]

    paired_summaries[k] = (v , the_summary, human_made)
```
